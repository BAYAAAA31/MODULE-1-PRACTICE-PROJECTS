```python
# numerical python (numpy) - numpy is one of python's core libraries used for scientific computing.
# for this module, it focusews more on numpy's objects and tools for manipulating arrays
# arrays - a data set that stores multiple items of the same data type under one name.

import numpy as np # this imports the library in the notebook

# ndarray: a multi-dimebnsional container of items of the same data type and size
# The number of dimensions and items in an array is defined by its shape, 
# which is a tuple of N non-negative integers that specify the sizes of each dimension.
```

```python
print ("numpy demo 1: creating and inspeacting ndarrays")

# i will showcase 3 different arrays: vectors(1D), matrices(2D), and tensors(2D)

# in this demo, we will be focusing on two functions: np.array and .shape
# example: 

vec = np.array([1,2,3]) # initializes an array
vec
#vec.shape  # shows the number of items of in an array (remove hashtag and this comment to show shape)
```

```python
print("for matrices, the .shape function shows the  number of rows and columns")
mat = np.array([(1,2,3),(4,5,6)])
mat
#mat.shape (remove hashtag and this comment to show shape)
```

```python
print("for tensors, the .shape function shows the  number of rows, columns, and layers of the array.")
ten = np.array([[(1,2,3,),(5,6,7)],[(8,9,10),(11,12,13)]])
ten
#ten.shape (remove hashtag and this comment to show shape)
```
