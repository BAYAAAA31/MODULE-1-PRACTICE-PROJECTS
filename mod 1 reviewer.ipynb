```python
print ("types of int and float operations")

x = float(input("enter you first number: "))
y = float(input("enter your second number: "))

addition = x + y
subtraction = x - y
multiplication = x * y
exponent = x ** y
division = x / y 
modulo = x // y 

print ("addition: ", addition)
print ("subtraction: ", subtraction)
print ("multiplication: ", multiplication)
print ("exponent: ", exponent)
print ("division: ", division)
print ("modulo: ", modulo)
```

```python
print("data structures")

phone = ["iphone", "xiaomi", "samsung"] # a data structure is an object that contains a number of objects aswell

print(phone[2]) #index numbers can be negative, setting the index number as 2 identifies as samsung, the case is he same when you
#enter -1 as the index number, it still is samsung (if index is negative, count backwards, and vice versa)

phone.append("honor") # append adds a new object in the list right after the end of an object
phone #declaring a list shows its contents
# running this syntax adds the object honor at the end, right next to samsung

phone.insert(2,"infinix") # insert is the same as append, but lets you place the object at a specific index location
phone
#the function will put infinix at index number 2, but since that particular undex number is occupied by samsung, 
#the function will move samsung and the following objects after it to the next index numnber

phone.remove("iphone") # this function removes an object from the list
phone

phone.sort() #sort function PERMANENTLY arranges the contents of your list in an ascending order,
#for strings: it arranges alpabetically from a to z. for int or float: lowest to largest number.
phone
#fun fact: it isnt mentioned in the module but yuou can actually reverse sort the list
#just simply type the argument "reverse=True" inside the parenthesis!
```

```python
print("data structures part 2: dictionary")

device = {"model":"xiaomi" , "battery":"5000mah" , "storage":"512gb"}
#dictionaries use curly braces, unlike lists, each data or object has an associated key-value to them, linked by semicolons

device.get("battery") #.get functions prints out the specified data

device["ram"] = "16gb" #this function adds a new object with an associated key-value to the dictionary
device

device.pop("battery") ## this function removes a data fromthe dictionary
device

device.popitem() ##this function is the same as the pop function, but does not take any arguments
## and it removes the very last data from the dictionary.
device
```
